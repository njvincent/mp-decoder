## Full scan
### Overall error rate

```python
# Full scan
from analysis_utils import (
    display_finite_size_scaling_fit,
    display_fit_summary,
    fit_finite_size_scaling,
)

# Both control and target fits use this scaling-polynomial degree.
fit_degree = 4

full_scan = load_ft_scan(
    "../results/cnot_primitive_consecutive/full_scan",
    p_values=[0.009, 0.010, 0.011, 0.012, 0.013, 0.014, 0.015, 0.016, 0.017, 0.018, 0.019],
    L_values=[5, 7, 9, 13, 19],
    expected_repeats=1,
    failure_fields={
        "target": "target_logical_failures",
        "control_1": "control_1_logical_failures",
        "control_2": "control_1_logical_failures"
    },
)
full_scan.print_summary()
```

```python
plot_rate_scan(
    full_scan,
    title="CNOT consecutive: T_CNOT_T_CNOT_T_2T",
    ylabel="error_rate",
)
```

### Control 1 error rate

```python
plot_rate_scan(
    full_scan,
    rate_name="control_1",
    title="CNOT consecutive control 1: T_CNOT_T_CNOT_T_2T",
    ylabel="control_1_error_rate",
)

fit_consecutive_control_1 = fit_finite_size_scaling(
    full_scan,
    rate_name="control_1",
    label="CNOT consecutive — control 1",
    polynomial_degree=fit_degree,
)
display_finite_size_scaling_fit(fit_consecutive_control_1, "control_1_error_rate")
```

### Control 2 error rate

```python
plot_rate_scan(
    full_scan,
    rate_name="control_2",
    title="CNOT consecutive control 2: T_CNOT_T_CNOT_T_2T",
    ylabel="control_1_error_rate",
)

fit_consecutive_control_2 = fit_finite_size_scaling(
    full_scan,
    rate_name="control_2",
    label="CNOT consecutive — control 2",
    polynomial_degree=fit_degree,
)
display_finite_size_scaling_fit(fit_consecutive_control_2, "control_2_error_rate")
```

### Target error rate

```python
plot_rate_scan(
    full_scan,
    rate_name="target",
    title="CNOT consecutive target: T/2_CNOT_T/2_2T",
    ylabel="target_error_rate",
)

fit_consecutive_target = fit_finite_size_scaling(
    full_scan,
    rate_name="target",
    label="CNOT consecutive — target",
    polynomial_degree=fit_degree,
)
display_finite_size_scaling_fit(fit_consecutive_target, "target_error_rate")
display_fit_summary(
    [fit_consecutive_control_1, fit_consecutive_control_2, fit_consecutive_target],
    title="CNOT consecutive fit summary",
)
```
